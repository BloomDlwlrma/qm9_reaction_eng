### Distribution of atimization energy of QM9 Database based on T-dnn

***The diagram shows the atomization energy distribution difference between osvccsd/6-31g and G4MP2 method.***


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import seaborn as sns
import logging
import matplotlib.font_manager as fm

fm.fontManager.addfont('/usr/share/fonts/truetype/times.ttf')
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['axes.unicode_minus'] = False

KCAL_TO_EV = 23.06054
```

#### Distribution of osvccsd/6-31g

```python
fig, ax1 = plt.subplots(figsize=(10, 8),dpi=1200)


mol_df_ccsd = pd.read_csv("/home/ubuntu/Shiwei/qm9_reaction_eng/csv/qm9_bonds_energies_ccsd_final.csv")
mol_df_ccsd = mol_df_ccsd.dropna(subset=["atomization_energy_osv", "atomization_energy_acc"])

mol_df_ev_ccsd = mol_df_ccsd.copy()
mol_df_ev_ccsd["atomization_energy_acc"] = mol_df_ev_ccsd["atomization_energy_acc"] / KCAL_TO_EV
mol_df_ev_ccsd["atomization_energy_osv"] = mol_df_ev_ccsd["atomization_energy_osv"] / KCAL_TO_EV


n_bins = 1000
xmin = min(mol_df_ev_ccsd["atomization_energy_osv"].min(), mol_df_ev_ccsd["atomization_energy_acc"].min()) 
xmax = max(mol_df_ev_ccsd["atomization_energy_osv"].max(), mol_df_ev_ccsd["atomization_energy_acc"].max()) 
for col in ["atomization_energy_osv", "atomization_energy_acc"]:
    s = mol_df_ev_ccsd[col]
    print(
        f"{col}: min={s.min():.6f} eV, max={s.max():.6f} eV, "
        f"neg={(s < 0).sum()}, pos={(s > 0).sum()}, zero={(s == 0).sum()}"
    )
print(f"combined range (min->max): {xmin:.6f} to {xmax:.6f} eV")


bin_edges = np.linspace(xmin, xmax, n_bins + 1)

sns.histplot(
    mol_df_ev_ccsd['atomization_energy_acc'], bins=bin_edges, kde=False, ax=ax1,
    color='#E34027', stat='count', element='step', fill=False, linewidth=1.2,
)
sns.histplot(
    mol_df_ev_ccsd['atomization_energy_osv'], label="osvccsd/6-31g; MAE = 0.43", bins=bin_edges, kde=False, ax=ax1,
    color='#EB9184', stat='count', alpha=0.8, edgecolor="white", linewidth=0.4,
)

# plt.xlabel("Atomization Energy (kcal/mol)")
plt.xlabel("Atomization Energy (eV)")
plt.ylabel("Count")
plt.title("Atomization Energy Distribution: osvccsd/6-31g", pad=15, fontsize=16)
plt.legend()
# plt.savefig("/home/ubuntu/Shiwei/qm9_reaction_eng/png/atomization_energy_distribution_comparison_ev.png", dpi=1200, bbox_inches='tight')
plt.xlim(-175, 0)
plt.savefig("/home/ubuntu/Shiwei/qm9_reaction_eng/png/atom_eng_distrib_ev_ccsd_n175_0.png", dpi=1200, bbox_inches='tight')
plt.show()
```

### Distribution of the error in reaction energies (eV) based on T-DNN methods for different functionals.

#### The diagram shows the error of T-dnn methods distribution difference between osvccsd/6-31g and osvmp2/6-31g method. The reaction energies mostly follow a normal distribution with the majority of reactions on a relatively small error scale.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import seaborn as sns
import matplotlib.font_manager as fm
target = ['ccsd', 'mp2']
```

#### Distribution of calculation error of osvccsd/6-31g

```python
# rxnindex,reactant_index,product_index,deltaE_osvmp2_631g,deltaE_Accurate_eng_mp2,deltaE_error
t = target[0] 

rxn_df_ccsd = pd.read_csv(f"/home/ubuntu/Shiwei/qm9_reaction_eng/csv/reactions_{t}.csv")

rxn_df_ccsd = rxn_df_ccsd.dropna(subset=[f"deltaE_osv{t}_631g", f"deltaE_Accurate_eng_{t}", "deltaE_error"])

# rxn_df_mp2 = rxn_dfs[t]
# rxn_df_mp2 = rxn_df_mp2.dropna(subset=[f"deltaE_osv{t}_631g", f"deltaE_Accurate_eng_{t}", "deltaE_error"])
```

```python
rxn_df_ccsd[f"deltaE_osvccsd_ev"] = rxn_df_ccsd[f"deltaE_osvccsd_631g"] / KCAL_TO_EV
rxn_df_ccsd[f"deltaE_Accurate_eng_ccsd_ev"] = rxn_df_ccsd[f"deltaE_Accurate_eng_ccsd"] / KCAL_TO_EV
rxn_df_ccsd[f"reaction_error_ccsd_ev"] = rxn_df_ccsd["deltaE_error"] / KCAL_TO_EV
```

```python
rxn_df_ccsd.head()
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

fig, ax = plt.subplots(figsize=(10, 8), dpi=1200)
sns.histplot(
    data=rxn_df_ccsd,
    x=f"reaction_error_ccsd_ev",
    bins=500,  
    kde=False,
    ax=ax,
    color="#EB9184",
    stat="count",
    alpha=0.8,
    edgecolor="white",
    linewidth=0.4,
    element="bars",
    fill=True,
    shrink=0.95,  
)

ax.set_xlabel("Error in Reaction Energies (eV)")
ax.set_ylabel("Counts")
ax.set_title(f"Distribution of Error in Reaction Energies osvccsd/6-31g")

legend_handles = [
    Patch(facecolor='#EB9184', edgecolor='white', label='osvccsd/6-31g; MAE = 0.43'),
    # Patch(facecolor='#80ACF9', edgecolor='white', label='osvmp2/6-31g; MAE = 0.44'),
]
ax.legend(handles=legend_handles)
# plt.xlim(-0.2, 0.2)

plt.tight_layout()
plt.savefig(f"/home/ubuntu/Shiwei/qm9_reaction_eng/png/react_eng_error_ccsd.png", dpi=1200, bbox_inches='tight')
# plt.savefig(f"/home/ubuntu/Shiwei/qm9_reaction_eng/png/react_eng_error_ccsd_n02_p02.png", dpi=1200, bbox_inches='tight')
plt.show()
```
